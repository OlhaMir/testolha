```python
# 1. Вхідні дані: два фактори та цільова змінна
x = np.array([
    [10., 2.],
    [12., 3.],
    [15., 4.],
    [18., 5.],
    [20., 6.]
])

y = np.array([21., 25., 31., 36., 40.])

# 2. Фільтрація
mask = np.isfinite(x).all(axis=1) & np.isfinite(y)
x = x[mask]
y = y[mask]

# 3. Стандартизація ознак
mu = x.mean(axis=0)
sigma = x.std(axis=0)
x_std = (x - mu) / sigma

# 4. Додаємо стовпець одиниць для вільного члена
x_design = np.column_stack([
    np.ones(x_std.shape[0]),
    x_std
])

# 5. Оцінювання параметрів методом найменших квадратів
beta, *_ = np.linalg.lstsq(x_design, y, rcond=None)

# 6. Прогноз і MSE
y_pred = x_design @ beta
mse = np.mean((y - y_pred) ** 2)

print("beta =", beta)
print("y_pred =", y_pred)
print("MSE =", mse)
```

```python
# 1.
np.random.seed(42)
X = np.random.randn(5, 4)

print("1. Матриця X:")
print(X)
print("Форма:", X.shape)


# 2.
mean = X.mean(axis=0)
std = X.std(axis=0)

print("\n2. Середнє кожної ознаки:")
print(mean)

print("Стандартне відхилення кожної ознаки:")
print(std)


# 3.
X_std = (X - mean) / std

print("\n3. Стандартизована матриця:")
print(X_std)


# 4.
positive_rows = X_std[X_std[:, 0] > 0]

print("\n4. Рядки, де перша ознака додатна:")
print(positive_rows)


# 5.
X_new = np.hstack((X_std, np.ones((X_std.shape[0], 1))))

print("\n5. Матриця з новим стовпцем одиниць:")
print(X_new)


# 6.
w = np.array([0.5, -0.3, 0.2, 0.7, 1.0])

y_pred = X_new @ w

print("\n6. Вектор ваг:")
print(w)

print("Прогноз X @ w:")
print(y_pred)


# 7.
y_true = np.array([1.0, 0.5, -0.5, 1.5, 0.0])

mse = np.mean((y_true - y_pred) ** 2)

print("\n7. y_true:")
print(y_true)

print("MSE:", mse)


# 8.
print("\n8. Перевірка масивів:")

print("Чи є NaN у X:", np.isnan(X).any())
print("Чи є inf у X:", np.isinf(X).any())

print("Чи є NaN у X_std:", np.isnan(X_std).any())
print("Чи є inf у X_std:", np.isinf(X_std).any())

print("Чи є NaN у прогнозі:", np.isnan(y_pred).any())
print("Чи є inf у прогнозі:", np.isinf(y_pred).any())
```
